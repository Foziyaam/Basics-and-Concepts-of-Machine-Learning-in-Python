Below are **step-by-step instructions** to install **Jupyter Notebook**, **NumPy**, **pandas**, and **matplotlib** on **Windows**, **macOS**, and **Linux**. The most straightforward approach is usually to install **Anaconda** or **Miniconda**, which comes with these packages (or makes installing them via `conda` easy). # 1. Using Anaconda (Recommended for All OS)

## 1.1. Why Anaconda?
- Bundles Python, Jupyter Notebook, and popular data science libraries (NumPy, pandas, matplotlib) in one installation.
- Manages environments easily to avoid dependency conflicts.

### 1.2. Download & Install
1. **Go to** [Anaconda Downloads](https://www.anaconda.com/products/distribution).
2. **Choose** your OS (Windows, macOS, or Linux).
3. **Download** the appropriate installer for Python 3.x (64-bit recommended).
4. **Run the installer**:
   - **Windows**: Double-click the `.exe` file and follow prompts.  
   - **macOS**: Double-click the `.pkg` file. You may need to allow permissions in System Preferences → Security & Privacy.  
   - **Linux**: Make the `.sh` file executable (`chmod +x`) and run it (`./Anaconda3-xxxx.sh`). Follow on-screen instructions.

### 1.3. Post-Install & Environment Check
- **Open the Anaconda Prompt** (Windows) or **Terminal** (macOS/Linux).
- **Check** conda version:
  ```bash
  conda --version
  ```
  If successful, you have Anaconda installed.

### 1.4. Installing/Updating Packages with Conda
- **Install** or **update** Jupyter, NumPy, pandas, matplotlib:
  ```bash
  conda install jupyter numpy pandas matplotlib
  ```
  (They might already be included in the base environment, but this ensures the latest versions.)

### 1.5. Launch Jupyter Notebook
- **From the Anaconda Prompt/Terminal**:
  ```bash
  jupyter notebook
  ```
- A browser tab should open showing the **Jupyter interface**.

# 2. Using Pip & System Python (Alternative)

## 2.1. Verify Python Installation

### Windows
1. **Open Command Prompt** or **PowerShell**.
2. Type:
   ```bash
   python --version
   ```
   If you see `Python 3.x.y`, Python is installed. If not, install from [python.org](https://www.python.org/downloads/windows/).

### macOS
- Modern macOS versions ship with Python 2 by default. You may need a **Python 3** install:
  1. **Go to** [python.org](https://www.python.org/downloads/macos/).
  2. **Download** the macOS 64-bit installer.
  3. **Double-click** the installer and follow the steps.
- **Check** Python version:
  ```bash
  python3 --version
  ```
  or
  ```bash
  python --version
  ```
  (depending on the path setup).

### Linux
- Most distros have Python 3 installed. Verify:
  ```bash
  python3 --version
  ```
  If not present, install from your distro’s package manager (e.g., `sudo apt-get install python3` on Ubuntu/Debian).

## 2.2. Ensure pip is Installed or Updated
- **Check** pip version:
  ```bash
  pip --version
  ```
  or (if pip is mapped to Python 2):
  ```bash
  pip3 --version
  ```
- If pip is missing, on Windows you can re-run the Python installer and select “Install pip.” On macOS/Linux, it often comes bundled with Python 3. You can install or upgrade pip by:
  ```bash
  python -m ensurepip --upgrade
  ```

## 2.3. Install Jupyter Notebook, NumPy, pandas, matplotlib with pip
### Windows
1. **Open Command Prompt** or **PowerShell**.
2. **Install** each package:
   ```bash
   pip install jupyter
   pip install numpy
   pip install pandas
   pip install matplotlib
   ```
   or do them in one line:
   ```bash
   pip install jupyter numpy pandas matplotlib
   ```
3. **Launch** Jupyter:
   ```bash
   jupyter notebook
   ```
   (If using Python 3 specifically, replace `pip` with `pip3`, if needed.)

### macOS
1. **Open Terminal**.
2. **Install**:
   ```bash
   pip3 install jupyter numpy pandas matplotlib
   ```
3. **Launch** Jupyter:
   ```bash
   jupyter notebook
   ```
4. If `jupyter` command not found, ensure `~/.local/bin` (or wherever pip installs scripts) is in your PATH.

### Linux
1. **Open Terminal**.
2. **Install**:
   ```bash
   pip3 install jupyter numpy pandas matplotlib
   ```
3. **Launch** Jupyter:
   ```bash
   jupyter notebook
   ```
   If permission issues appear, you might prepend `sudo` or consider using a virtual environment.

# 3. Detailed OS-Specific Notes & Troubleshooting

## 3.1. Windows Specifics
- **Anaconda Navigator**: Windows users can open a GUI called “Anaconda Navigator” to launch Jupyter Notebook, manage environments, etc.
- **PATH Settings**: If `python` or `jupyter` is not recognized, ensure that during installation you chose the option to add Python/Anaconda to the PATH, or re-run the installer and modify the path.
- **PowerShell Execution**: If you get execution policy errors, you may need to set `RemoteSigned` or `Bypass` policies for scripts.

## 3.2. macOS Specifics
- macOS might show a security prompt about opening an app from an “unidentified developer” when installing Python or Anaconda. You can override by going to **System Preferences → Security & Privacy** and clicking “Open Anyway.”
- **Homebrew** alternative: `brew install python` or `brew cask install anaconda` (though cask might be outdated). Then proceed with pip or conda.

## 3.3. Linux Specifics
- You can also install Python 3, pip, and scientific libraries from your distro’s package manager:
  ```bash
  sudo apt-get update
  sudo apt-get install python3 python3-pip python3-dev
  ```
  Then `pip3 install jupyter numpy pandas matplotlib`.
- For advanced usage, you can create virtual environments with `python3 -m venv envname` or use `conda` if Anaconda is installed.

### **Slide 33: Introduction to Linear Regression**
- **Definition:**
  - Linear regression predicts a continuous dependent variable based on one or more independent variables.
  - The relationship is modeled as a straight line: \( y = \beta_0 + \beta_1x_1 + \epsilon \).
- **Use Case:**
  - Predicting house prices based on size, location, and number of rooms.
- **Key Concepts:**
  - Coefficients (\( \beta_0, \beta_1 \)): Represent the slope and intercept.
  - Residuals: Differences between actual and predicted values.
- **Python Code Example:**

````python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.1, 4.1, 6.1, 8.1, 10.1])
model = LinearRegression()
model.fit(X, y)
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
````

- **Insights:**
  - Red line represents the best-fit line.

### **Slide 35: Logistic Regression**
- **Definition:**
  - Logistic regression is used for binary classification tasks.
  - The output is modeled using a sigmoid function: \( P(y=1) = \frac{1}{1+e^{-z}} \), where \( z = \beta_0 + \beta_1x_1 \).
- **Use Case:**
  - Predicting whether an email is spam or not.
- **Python Code Example:**

````python
from sklearn.linear_model import LogisticRegression

X = [[0], [1], [2], [3], [4]]
y = [0, 0, 1, 1, 1]
model = LogisticRegression()
model.fit(X, y)
print("Predictions:", model.predict([[1.5], [3.5]]))
````

### **Slide 36: Logistic Regression Visualization**

### **Slide 15: Mean Absolute Error (MAE)**
- Measures the average absolute difference between actual and predicted values.

- **Interpretation:**
  - Lower MAE = Better model performance.
  - Does not penalize large errors as much as MSE.

---

### **Slide 16: Mean Squared Error (MSE)**
- Measures the average squared difference between actual and predicted values.

- **Interpretation:**
  - Penalizes larger errors more than MAE.
  - Sensitive to outliers.
 
**Python Code Example:**

````python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
````

### **Slide 17: Root Mean Squared Error (RMSE)**
- RMSE is simply the square root of MSE.


- **Interpretation:**
  - Expresses error in the same units as the target variable.

**Python Code Example:**

````python
import numpy as np
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error:", rmse)
````

---

### **Slide 18: R² Score (Coefficient of Determination)**
- Measures the proportion of variance explained by the model.



**Python Code Example:**

````python
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)
````

- **Steps:**
  1. Load dataset.
  2. Normalize images.
  3. Build a simple CNN model.

````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Load dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0  # Normalize

# Build CNN model
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
````

## Slide 2: Introduction to Support Vector Machines (SVMs)
- **Definition of SVMs:** Support Vector Machines (SVMs) are supervised learning models that analyze data for classification and regression tasks. They work by identifying the best boundary (or hyperplane) that separates different classes in the data.
- **Key Idea:** The core concept of SVMs is finding the *optimal hyperplane* that maximally separates data points of different classes. This boundary is determined based on the data points closest to the hyperplane, known as *support vectors*.
- **Example:** Visualizing linearly separable data where a straight line divides two distinct classes clearly.

## Slide 3: The Mathematics Behind SVMs
- **Concept of Hyperplanes:** A hyperplane is a flat affine subspace of one dimension less than its ambient space. In 2D, it's a line; in 3D, it's a plane. SVMs aim to find the hyperplane that best separates different classes.
- **Definition of Margins and Support Vectors:** The *margin* is the distance between the hyperplane and the closest data points from each class. These closest points are called *support vectors*, and they are critical in defining the optimal hyperplane.
- **Maximizing the Margin:** SVMs work by maximizing the margin, which leads to better generalization of the model on new data. A larger margin reduces the risk of overfitting.

## Slide 4: Linear SVM Classifier
- **Linear Decision Boundary:** In binary classification, the decision boundary separates two classes using a straight line (or hyperplane in higher dimensions). This boundary is defined by the weights and bias learned during training.
- **Mathematical Formulation:** The decision function is given by \( w.x + b = 0 \), where \( w \) is the weight vector, \( x \) is the input feature vector, and \( b \) is the bias term. The goal is to find \( w \) and \( b \) that maximize the margin while correctly classifying the data.
- **Example:** Implementing a basic linear SVM in Python.

````python
from sklearn.svm import SVC
from sklearn.datasets import make_classification
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

# Generating sample data
X, y = make_classification(n_samples=100, n_features=2, n_classes=2, n_clusters_per_class=1) # X is the data and y is label (iris X, y)

# Linear SVM model
svm_model = SVC(kernel='linear')
svm_model.fit(X, y)

# Plotting the decision boundary
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr')
plt.title('Linear SVM Decision Boundary')
plt.show()
````

## Slide 5: Non-Linear SVM and the Kernel Trick
- **Introduction to the Kernel Trick:** When data isn't linearly separable, the kernel trick helps by transforming data into a higher-dimensional space where a linear separator can be found. This transformation is done implicitly without increasing computational complexity.
- **Common Kernels:**
  - *Polynomial Kernel:* Captures interactions of features up to a specified degree.
  - *Radial Basis Function (RBF) Kernel:* Measures similarity between points using Gaussian distribution.
- **When to Use Kernelized SVMs:** Kernel SVMs are preferred when data shows complex, non-linear patterns that cannot be separated by a straight line.

## Slide 6: SVM Parameters and Their Impact
- **Key Parameters:**
  - `C`: Controls the trade-off between achieving a low training error and a low testing error. A small `C` allows for a wider margin but may misclassify more points. A large `C` aims for fewer misclassifications but may overfit.
  - `gamma`: Determines the influence of a single training example. A low value considers points far from the decision boundary, while a high value focuses on points closer to it.
  - `kernel`: Specifies the type of kernel function used ('linear', 'poly', 'rbf', etc.).
- **Impact of Parameters:** Proper tuning of `C` and `gamma` is critical for model performance. Cross-validation is often used to find the optimal combination.

## Slide 7: Implementing SVM with RBF Kernel
- **Example Using RBF Kernel:** Applying the SVM model with an RBF kernel to handle non-linearly separable data, such as the `make_moons` dataset.

````python
from sklearn.svm import SVC
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt

# Generating non-linear data
X, y = make_moons(n_samples=100, noise=0.2)

# RBF Kernel SVM
svm_rbf = SVC(kernel='rbf', C=1, gamma=0.5)
svm_rbf.fit(X, y)

# Plotting the decision boundary
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
plt.title('SVM with RBF Kernel')
plt.show()
````

- **Explanation:** The RBF kernel captures the non-linear patterns in the data, enabling the SVM to classify points that a linear kernel cannot handle effectively.

## Slide 14: Cross-Validation Techniques

- **K-Fold Cross-Validation:**
  - Splits the dataset into *K* equal-sized folds.
  - The model is trained on *K-1* folds and tested on the remaining fold.
  - This process is repeated *K* times, with each fold serving as the test set once.
  - The performance metrics from each fold are averaged to provide a robust estimate of model performance.
  - **Python Example:**

````python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.svm import SVC
import numpy as np

X = np.random.rand(100, 5)
y = np.random.randint(0, 2, 100)

kf = KFold(n_splits=5)
model = SVC(kernel='linear')

scores = cross_val_score(model, X, y, cv=kf)
print(f'Average Accuracy: {scores.mean()}')
````

- **Stratified K-Fold for Imbalanced Datasets:**
  - Ensures each fold maintains the same class distribution as the original dataset.
  - Reduces bias in performance estimates, especially for imbalanced classification problems.
  - **Python Example:**

````python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=5)
scores = cross_val_score(model, X, y, cv=skf)
print(f'Average Accuracy with Stratified K-Fold: {scores.mean()}')
````

- **Leave-One-Out Cross-Validation (LOOCV):**
  - Each sample in the dataset serves as the test set exactly once, while the remaining samples form the training set.
  - Ideal for small datasets but computationally expensive for large datasets.
  - **Python Example:**

````python
from sklearn.model_selection import LeaveOneOut

loo = LeaveOneOut()
scores = cross_val_score(model, X, y, cv=loo)
print(f'Average Accuracy with LOOCV: {scores.mean()}')
````

## Slide 16: Advantages and Limitations of SVMs

- **Advantages:**
  - **Effective in High-Dimensional Spaces:** SVMs perform well when the number of dimensions is greater than the number of samples, making them suitable for text classification, gene expression data, etc.
  - **Versatile with Kernels:** The kernel trick allows SVMs to solve non-linear problems without explicitly increasing computational complexity.
  - **Robust to Overfitting:** Especially when using regularization parameters like `C`, SVMs manage overfitting effectively, even with noisy datasets.

- **Limitations:**
  - **Computationally Intensive:** Training time can be high for large datasets because SVMs rely on solving quadratic optimization problems.
  - **Sensitive to hyperparameter Tuning:** The performance of SVMs heavily depends on the choice of kernel, regularization parameters (`C`), and kernel-specific parameters like `gamma`.
  - **Poor Performance with Overlapping Classes:** SVMs may struggle with datasets where classes are heavily overlapped, as the model aims for maximum margin, which might not be meaningful in such cases.

- **Real-World Use Cases:**
  - **Text Categorization:** Classifying emails as spam or not spam using linear SVMs.
  - **Image Classification:** Recognizing handwritten digits (like MNIST dataset) using kernelized SVMs.

## Slide 18: Handling Imbalanced Data

## Slide 26: Model Evaluation Metrics

## Slide 28: ROC Curve and AUC

- **Understanding Receiver Operating Characteristic (ROC) Curve:**
  - A graphical plot illustrating the diagnostic ability of a binary classifier.
  - Plots **True Positive Rate (TPR)** against **False Positive Rate (FPR)** at various threshold settings.
  - Helps visualize the trade-off between sensitivity (recall) and specificity.

- **Area Under the Curve (AUC) as a Performance Measure:**
  - AUC quantifies the overall ability of the model to discriminate between classes.
  - **AUC = 1**: Perfect model, **AUC = 0.5**: No discrimination (random guessing).
  - Higher AUC indicates a better performing model.

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

X, y = make_classification(n_samples=1000, n_classes=2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = SVC(probability=True)
model.fit(X_train, y_train)
y_scores = model.predict_proba(X_test)[:, 1]

fpr, tpr, _ = roc_curve(y_test, y_scores)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
````

## Slide 30: Cross-Validation Techniques

- **k-Fold Cross-Validation:**
  - Splits data into *k* equal parts (folds).
  - Trains on *k-1* folds and tests on the remaining fold, rotating folds.
  - Reduces bias and variance by ensuring every data point is used for both training and validation.

- **Stratified k-Fold for Imbalanced Datasets:**
  - Ensures each fold maintains the same class distribution as the full dataset.
  - Critical for classification problems with imbalanced classes.

- **Leave-One-Out Cross-Validation (LOOCV):**
  - Special case of k-Fold where *k* equals the number of samples.
  - Trains on *n-1* samples and tests on 1 sample.
  - High computational cost but useful for small datasets.

````python
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from sklearn.datasets import load_iris
from sklearn.svm import SVC

iris = load_iris()
X, y = iris.data, iris.target

# k-Fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(SVC(), X, y, cv=kf)
print("k-Fold Accuracy:", scores.mean())

# Stratified k-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
stratified_scores = cross_val_score(SVC(), X, y, cv=skf)
print("Stratified k-Fold Accuracy:", stratified_scores.mean())
````

## **Slide 7: Backpropagation**
- **Purpose**: Algorithm to train neural networks by adjusting weights to minimize error.
- **Process**:
  1. Forward pass to compute prediction.

````python
class MyCustomLayer(keras.layers.Layer):
    def __init__(self, units=32):
        super().__init__()
        self.units = units

    def build(self, input_shape):
        self.w = self.add_weight(shape=(input_shape[-1], self.units))

    def call(self, inputs):
        return tf.matmul(inputs, self.w)
````

# Module 10: Deep Learning for Computer Vision (CNNs)

---

## **Slide 1: Introduction to Computer Vision and CNNs**
- **What is Computer Vision?**  
  - A field of AI focused on enabling machines to interpret and understand visual data.
- **Why Convolutional Neural Networks (CNNs)?**  
  - Designed to process image data by capturing spatial relationships with minimal preprocessing.
- **Historical Milestones**:  
  - LeNet (1990s) → AlexNet (2012) → ResNet/VGG (2014–2015) → Transformers (recent).
- **Objective of Module**:  
  - Learn CNN architecture fundamentals, implement them in Python with Keras.

---

## **Slide 2: Traditional Image Processing vs. Deep Learning**
- **Traditional Approach**:  
  - Handcrafted features (e.g., edge detectors, SIFT, HOG).  
  - Limited to domain expertise, less flexible.
- **Deep Learning Approach**:  
  - CNNs learn hierarchical features automatically.  
  - Highly scalable with data and computing power.
- **Applications**:  
  - Image classification, object detection, segmentation, face recognition.

---

## **Slide 3: CNN Core Concepts**
- **Convolution Operation**:  
  - Uses filters (kernels) to detect local patterns like edges, corners.
- **Local Receptive Fields**:  
  - Neurons only connect to a small region of the input, capturing spatial dependencies.
- **Shared Weights**:  
  - Each filter is reused across the entire input image, reducing the number of parameters.

---

## **Slide 4: Convolution Operation Explained**
- **Filter (Kernel)**:  
  - A small matrix of weights that slides over the image.
- **Stride and Padding**:  
  - **Stride**: Controls how many pixels the filter moves at each step.  
  - **Padding**: Preserves the spatial size of the image by adding zeros around edges.
- **Mathematical Example**:  
  - Output pixel = Summation of (Kernel Element * Image Patch Element).

---

## **Slide 5: Pooling Layers**
- **Purpose**:  
  - Reduce spatial dimensions, control overfitting.
- **Types**:
  - **Max Pooling**: Takes the maximum value from each patch.
  - **Average Pooling**: Takes the average value from each patch.
- **Illustration**:  
  - 2×2 max pooling reduces dimension by half in both width and height.

---

## **Slide 6: CNN Architecture Overview**
- **Common Structure**:  
  1. **Convolution Layer(s)**  
  2. **Pooling Layer(s)**  
  3. **Fully Connected Layer(s)**  
  4. **Output Layer** (softmax for classification)
- **Feature Hierarchy**:  
  - Early layers learn edges, lines; deeper layers learn shapes, objects.

---

## **Slide 7: Activation Functions in CNNs**
- **ReLU**:  
  - Most common due to faster training and mitigating vanishing gradients.
- **Leaky ReLU** or **Parametric ReLU**:  
  - Help address “dying ReLU” problem.
- **Sigmoid/Tanh**:  
  - Rarely used in hidden layers for CNNs, can saturate and slow training.

---

## **Slide 8: Example – Simple CNN with Keras**

````python
import tensorflow as tf
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
````

- **Explanation**:
  - First Conv layer: 32 filters, each 3×3.
  - MaxPooling reduces spatial dimension by half.
  - Flatten transforms 2D to 1D.
  - Dense layers for classification.

---

## **Slide 9: Why Convolution?**
- **Spatial Invariance**:  
  - Objects recognized despite location changes.
- **Fewer Parameters**:  
  - Shared weights drastically reduce parameter count vs. fully connected layers on images.
- **Translation Equivariance**:  
  - Shifting input shifts output correspondingly.

## **Slide 5: K-Means Clustering Overview**
- **Concept**:
  - Assign points to k clusters by minimizing within-cluster variance.
- **Steps**:
  1. Initialize k centroids.
  2. Assign each point to nearest centroid.
  3. Update centroids as mean of assigned points.
  4. Repeat until convergence.
- **Pros/Cons**:
  - **Pros**: Fast, easy to implement.
  - **Cons**: Needs k, sensitive to outliers, spherical cluster assumption.

---

## **Slide 6: K-Means Algorithm Steps in Detail**
1. **Choose k**: The number of clusters.
2. **Random Initialization**: Place k centroids randomly.
3. **Assignment Step**: For each point, find centroid with min distance.
4. **Update Step**: Recompute centroid as average of assigned points.
5. **Convergence Check**: Stop if assignments no longer change or max iterations reached.

---

## **Slide 15: DBSCAN Overview**
- **Density-Based Spatial Clustering**:
  - Defines clusters as high-density regions separated by low-density areas.
- **Parameters**:
  - **eps**: Neighborhood radius.
  - **min_samples**: Minimum points to form a dense region.
- **Advantages**:
  - Finds arbitrarily shaped clusters.
  - Identifies outliers as noise.

---

## **Slide 16: DBSCAN Algorithm Steps**
1. For each point, check how many neighbors within `eps`.
2. If >= `min_samples`, becomes a “core point,” forming cluster.
3. Points within range of a core point → belong to same cluster.
4. Points not reachable → labeled outliers.

---

## **Slide 17: Code Example – DBSCAN**

````python
from sklearn.cluster import DBSCAN
db = DBSCAN(eps=0.5, min_samples=5)
labels = db.fit_predict(X)
print("Cluster Labels:", labels)
````

- **Label** `-1` indicates outlier.

---

## **Slide 18: Handling Outliers with DBSCAN**
- **Noise Points**: 
  - DBSCAN isolates outliers as separate label -1.
- **Parameters Tuning**:
  - `eps` too small → many outliers, 
  - `eps` too large → merges clusters.
- **Grid search** approach or domain knowledge.

## **Slide 19: Gaussian Mixture Models (GMM)**
- **Concept**:
  - Assumes data from a mixture of Gaussian distributions.
- **EM Algorithm**:
  1. **E-step**: Compute responsibility (probability of belonging to each Gaussian).
  2. **M-step**: Update Gaussian parameters (mean, covariance, mixture weight).
- **Pros**: Soft clustering, cluster probabilities.

---

## **Slide 20: GMM Example Code**

````python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=2, random_state=42)
gmm.fit(X)
labels = gmm.predict(X)
probs = gmm.predict_proba(X)
print("Cluster labels:", labels)
print("Probabilities:", probs[:5])
````

- **Explanation**: We get both discrete cluster labels and membership probabilities.

## **Slide 4: Recap – Supervised Learning**
- **Definition**: Learning a function from labeled examples.
- **Common Tasks**:
  - Regression (predict continuous values).
  - Classification (predict discrete classes).
- **Typical Workflows**:
  1. Train-test split.
  2. Model training.
  3. Hyperparameter tuning (CV).
  4. Evaluation metrics (RMSE, accuracy, F1).

---

## **Slide 5: Regression Methods Reviewed**
- **Linear Regression**:
  - OLS for simplicity, can be extended (Ridge, Lasso).
- **Decision Tree Regression**:
  - Non-linear, interpretable, can overfit.
- **Random Forest**:
  - Ensemble of trees, robust to outliers, better generalization.
- **Metrics**:
  - MSE, RMSE, MAE, R².

---

## **Slide 6: Classification Methods Reviewed**
- **Logistic Regression**: Good baseline, interpretable coefficients.
- **Support Vector Machines**: Effective for medium-size data, kernel transformations for non-linear boundaries.
- **Decision Trees & Ensembles**: Bagging (Random Forest), Boosting (XGBoost, Gradient Boosted Trees).
- **Metrics**:
  - Accuracy, precision, recall, F1, AUC.

---

## **Slide 7: Model Evaluation & Cross-Validation**
- **Cross-Validation**: K-fold, Stratified K-fold for classification.
- **Why**: More robust measure than a single train/test split.
- **Hyperparameter Tuning**:
  - GridSearchCV, RandomizedSearchCV.
- **Pitfall**: Data leakage must be avoided (preprocessing within cross-validation).

---

## **Slide 8: Recap – Unsupervised Learning**
- **Clustering**: K-means, DBSCAN, Hierarchical.
- **Dimensionality Reduction**: PCA, t-SNE.
- **Applications**: Customer segmentation, anomaly detection, discovering hidden structures.

## **Slide 9: Recap – Convolutional Neural Networks (CNNs)**
- **Key Idea**: Convolution layers for spatial data, particularly images.
- **Typical Architecture**: Conv → Pool → Dense.
- **Use Cases**: Image classification, object detection, segmentation.
- **Modern Variants**: ResNet, VGG, MobileNet.

## **Slide 10: Recap – Recurrent Neural Networks (RNNs)**

- **Vanishing Gradients**: LSTM, GRU gating solve it.
- **Applications**: Language modeling, sentiment analysis, forecasting.

## **Slide 15: Revisited – Overfitting vs Underfitting**
- **Overfitting**: High training score, low test score. Solutions: Regularization, more data, simpler model.
- **Underfitting**: Both training & test scores are poor. Solutions: Increase model complexity or gather more relevant features.

## **Slide 21: GPU vs. CPU in Production**
- **Tradeoffs**:

- **Decision**: Evaluate concurrency, response time needs, cost constraints.

- **Monitoring**: Periodic re-check of performance, feature distributions.
- **Remediation**: Retrain or adapt model on new data.
